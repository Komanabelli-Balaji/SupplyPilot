```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("../reports/baseline.csv")
```

## Bullwhip

```python
plt.figure(figsize=(12,5))
plt.plot(df.day, df.bullwhip_ratio)
plt.title("Bullwhip Effect")
plt.xlabel("Simulation Day")
plt.ylabel("Bullwhip Ratio")
plt.grid()
plt.show()
```

## Service level

```python
plt.figure(figsize=(12,5))
plt.plot(df.day, df.service_level)
plt.grid()
plt.show()
```

## Retailer profit

```python
plt.figure(figsize=(12,5))
plt.plot(df.day, df.retailer_profit)
plt.title("Daily Retailer Profit")
plt.xlabel("Simulation Day")
plt.ylabel("Profit")
plt.grid()
plt.show()
```

## Distributor profit

```python
plt.figure(figsize=(12,5))
plt.plot(df.day, df.distributor_profit)
plt.title("Daily Distributor Profit")
plt.xlabel("Simulation Day")
plt.ylabel("Profit")
plt.grid()
plt.show()
```

## Factory profit

```python
plt.figure(figsize=(12,5))
plt.plot(df.day, df.factory_profit)
plt.title("Daily Factory Profit")
plt.xlabel("Simulation Day")
plt.ylabel("Profit")
plt.grid()
plt.show()
```

## Holding costs

```python
plt.figure(figsize=(12,5))
plt.plot(df.day, df.retailer_holding_cost,label="Retailer")
plt.plot(df.day, df.distributor_holding_cost,label="Distributor")
plt.plot(df.day, df.factory_holding_cost,label="Factory")

plt.title("Daily Holding Costs Across Supply Chain")
plt.xlabel("Simulation Day")
plt.ylabel("Holding Cost")

plt.legend()
plt.grid()
plt.show()
```
